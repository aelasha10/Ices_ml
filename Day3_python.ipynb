range(10) =>0-9
if we write range(10) it will print 10 only
if we add list then it will print in the form of list list(range(10))

```python
import time

list1 = list(range(10))
total = 0   # don't use 'sum' as a variable name

# using python list
start_time = time.time()

for num in list1:
    total += num

end_time = time.time()

print(f"Sum of elements: {total}")
print(f"Time taken: {end_time - start_time} seconds")
```
